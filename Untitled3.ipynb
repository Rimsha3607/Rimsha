```python

import matplotlib.pyplot as plt
import numpy as np
data = np.random.randn(1000)   # 1000 random numbers (normal distribution)

plt.hist(data, bins=30)

plt.title("Simple Histogram Example")
plt.xlabel("Values")
plt.ylabel("Frequency")

plt.show()
```
